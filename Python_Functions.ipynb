```python
def check_even_odd(num):
    if num % 2 == 0:
        print("EVEN")
    else:
        print("ODD")
```

```python
def max_number(a,b,c):
    if a > b and a > c:
        print(a,"Is Greater")
    elif b > a and b > c:
        print(b,"Is Greater")
    elif c > a and c > b:
        print(c,"Is Greater")
    else:
        print("Try Again")
```
