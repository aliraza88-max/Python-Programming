## DATA STRUCTURES
* Lists:Ordered,Mutable,Allows Duplicate Elements

```python
a=["a","b","c","d","e"]
print(a[0])  #accessing the first element of the list
print(a[1])  #accessing the second element of the list
```

* List Methods

```python
a.append("f")  # adding an element to the end of the list
a.insert(2,"f") # adding an element at a specific index
a.remove("b") # removing an element from the list
a.pop() # removing the last element from the list
a.sort() # sorting the list in ascending order
```

```python
a
```

```python
for a in a:
    print(a)
```
