```python
def sum(a,b):
    return a+b

def sub(a,b):
    return a-b

def mul(a,b):
    return a*b

def div(a,b):
    return a/b
```

```python
num_1=int(input("Enter 1st Number:"))
num_2=int(input("Enter 2nd Number:"))
operation=input("Enter operation(+,-,/,*):")

if operation == "+":
    print(num_1 + num_2)

elif operation == "-":
    print(num_1 - num_2)

elif operation == "/":
    print(num_1 / num_2)

elif operation == "*":
    print(num_1 * num_2)

else:
    print("Try Again")
```
